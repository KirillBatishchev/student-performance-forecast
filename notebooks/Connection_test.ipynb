```python
import dvc.api
from dvc.api import DVCFileSystem
import pandas as pd

REPO_PATH = '..' # Путь к корню проекта (шаг назад из notebooks)
fs = DVCFileSystem(REPO_PATH) # Открываем систему файлов DVC репозитория

all_files = fs.find("/KT3", detail=False, dvc_only=True) # Находим нашу папку с данными

'''
Вот тут есть нюанс. Путь получился максимально дегродский: '/KT3/KT3/u1.csv',
но, в целом, житьс этим можно.
'''
if bool(len(all_files)):
    with dvc.api.open(all_files[0], repo=REPO_PATH) as f: # Открываем первый файлик
        data = pd.read_csv(f)
    data.display()
```
